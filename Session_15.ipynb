```python
import pandas as pd


orders = pd.DataFrame({
   "order_id": [101, 102, 103, 104, 105, 106, 107, 108],
   "customer_id": [1, 2, 1, 3, 2, 1, 4, 3],
   "product": ["Notebook", "Pen Set", "Desk Lamp", "Backpack", "Mouse", "Notebook", "Chair", "Pen Set"],
   "amount": [45, 120, 899, 1299, 350, 45, 560, 120],
   "city": ["Bengaluru", "Mumbai", "Bengaluru", "Chennai", "Mumbai", "Bengaluru", "Pune", "Chennai"]
})


customers = pd.DataFrame({
   "customer_id": [1, 2, 3, 5],
   "customer_name": ["Ananya", "Rahul", "Priya", "Vikram"],
   "segment": ["Retail", "Retail", "Wholesale", "Retail"]
})


print(orders)
print(customers)
```

```python
a = orders.groupby("product")["amount"].sum()
print(a)
```

```python
a = orders.groupby("city")["amount"].max()
print(a)
```

```python
print(orders.groupby("city")["order_id"].sum())
```

```python
# dataframe.groupby(“column”)[“value_column”].agg([“sum”, “mean”, “count”])


a = orders.groupby('city')['amount'].agg(["sum","mean"])
print(a)
```

```python
#group by with multiple columns
a = orders.groupby(["city","product"])["amount"].sum()
print(a)
```

```python
# combining tables

#left_dataframe.merge(right_dataframe, on=”shared_column”, how=”inner”)
inner = orders.merge(customers,on = "customer_id",how="inner")
print(inner)
```

```python
print(inner[["order_id","customer_id","customer_name","amount"]])
```

```python
left_join = orders.merge(customers,left_on="customer_id",right_on="customer_id",how="left")
print(left_join)
```

```python
_cross_join = orders.merge(customers,how="cross")
print(_cross_join)
print(_cross_join.shape)
```

```python
merged = orders.merge(customers,on="customer_id",how= "left")
# print(merged)

segment_summary = merged.groupby("segment",dropna= False)["amount"].sum().reset_index()
print(segment_summary)
```

```python
print(orders['amount'].sum())
print(segment_summary['amount'].sum())
```

```python
a = orders.groupby('city')['amount'].agg(["sum","mean"])

b = a.sort_values("sum",ascending=False)
print(b)

```
