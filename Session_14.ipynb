```python
import pandas as pd
import numpy as np

```

```python



data = {
   "name": [" ananya rao", "RAHUL Verma", "Priya Nair", "Priya Nair", "sana khan", "Vikram  Singh", "Karan Mehta", "meera IYER", "Arjun Das", None],
   "city": ["Bengaluru", "mumbai", "Chennai", "Chennai", "PUNE", "bengaluru", "Mumbai", "chennai", "Pune", "Bengaluru"],
   "math_score": [78, 55, 92, 92, np.nan, 64, 71, 95, np.nan,49]
}


raw_students = pd.DataFrame(data)
print(raw_students)
```

```python
print(raw_students.isnull())
```

```python
# dataframe.isnull().sum.......first commond
print(raw_students.isnull().sum())
```

```python
print(raw_students["math_score"].isnull())
```

```python

print(raw_students["math_score"].isnull().sum())
```

```python
print(raw_students["city"].isnull().any())
```

```python
print(raw_students.isnull().sum().sum())
```

```python
dropped = raw_students.dropna()
print(dropped)
print("rows before",len(raw_students))
print("rows after",len(dropped))
# droping....................................
```

```python
#filling the null value
avg_scr = raw_students["math_score"].mean()
print(avg_scr)

filled = raw_students.copy()
print(filled)

filled["math_score"]=filled["math_score"].fillna(avg_scr)
print(filled)
```

```python
print(raw_students.duplicated())
```

```python
a  = raw_students.drop_duplicates()
print(a)
print("row before:",len(raw_students))
print("row after:",len(a))
```
