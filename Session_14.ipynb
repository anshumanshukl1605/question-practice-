```python
import pandas as pd
import numpy as np

data = {
    "name": ["  ananya rao ", "RAHUL   VERMA", "  ananya rao ", None, "Priya Nair"],
    "city": ["delhi", "MUMBAI", "delhi", "PUNE", " Mumbai "],
    "math_score": ["80", "90", "80", "70", None]
}

df = pd.DataFrame(data)
print(df.isnull().sum())
df["math_score"]=df["math_score"].astype(float)
a = df["math_score"].mean()
df["math_score"] =df ["math_score"].fillna(a)
print(df["math_score"])
#step 3
df["name"] = df["name"].fillna("Unknown").str.strip().str.split().str.join(" ").str.title()

print(df["name"])
#step_4
df["city"]= df["city"].str.strip().str.title()
print(df["city"].unique())
#step_5
df["math_score"] = df["math_score"].astype(int)
print(df["math_score"].dtypes)
#step_6
print(df.duplicated().sum())
cleaned_df = df.drop_duplicates()
print(len(df))
print(len(cleaned_df))
```

```python
import pandas as pd
import numpy as np

data = {
    "name": ["  ananya rao ", "RAHUL   VERMA", "  ananya rao ", None, "Priya Nair"],
    "city": ["delhi", "MUMBAI", "delhi", "PUNE", " Mumbai "],
    "math_score": ["80", "90", "80", "70", None]
}

df = pd.DataFrame(data)

print(df)
print(df.isnull().sum())
df["math_score"]=df["math_score"].astype(float)
a = df["math_score"].mean()
df["math_score"] =df ["math_score"].fillna(a)
print(df["math_score"])

df["name"] = df["name"].fillna("Unknown").str.strip().str.split().str.join(" ").str.title()

print(df["name"])

df["city"]= df["city"].str.strip().str.title()
print(df["city"].unique())

df["math_score"] = df["math_score"].astype(int)
print(df["math_score"].dtypes)

print(df.duplicated().sum())
cleaned_df = df.drop_duplicates()
print(len(df))
print(len(cleaned_df))
```
