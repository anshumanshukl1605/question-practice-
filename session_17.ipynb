```python
# What is a probability of heads showing up when u toss a coin

np.random.seed(45)
for n in [20, 200, 20000]:
  flips = np.random.randint(0,2,n) # 0 = tails, 1 = heads
  heads = (flips == 1).sum()
  print(heads/n)
```

```python
space = []
for i in range(1, 7): # takes the first dice, one value at a time from 1 to 6
  for j in range(1, 7): # for each of those i, take the second dice from 1 to 6
    space.append((i,j)) # write downs the pairs

print(space)
print(len(space))
```

```python
# what is the chance that the two dice add up to 7 ? And what is the chance
# they don't add up to 7 ?


seven = []

for i in space:
  if i[0] + i[1] == 7:
    seven.append(i)

print(seven)

print(len(seven) / len(space))
```

```python
# out of our 20 orders, how many are from a NEW customer AND were LATE ? What
# is the probability , Both things must be true and the same times

# & mean AND (both must be true)
# | means OR atleast one of the conditions
# ~ means NOT, ~orders["on_time"] means "not on time"

new_and_late = orders[(orders["customer_type"] == "new") & (~orders["on_time"])]
#print(new_and_late)

print(len(new_and_late) / len(orders))
```

```python
new_and_late = orders[(orders["customer_type"] == "new") | (~orders["on_time"])]
#print(new_and_late)

print(len(new_and_late) / len(orders))
```
