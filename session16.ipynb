```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


sales = pd.DataFrame({
   "month": ["Jan", "Feb", "March", "Apr", "May", "June"],
   "revenue": [120, 135, 128, 160, 175, 190],
   "costs": [90, 95, 100, 110, 120, 125]
})


print(sales)
```

```python
fig,ax = plt.subplots()
ax.plot(sales["month"],sales["revenue"])
plt.show()
```

```python
# add a profit column (revenue - costs) to sales and draw it as a  line chart
sales ["profit"] = sales["revenue"]- sales["costs"]
fig, ax = plt.subplots()
ax.plot(sales["month"],sales["profit"])
plt.show()
```

```python
sales ["profit"] = sales["revenue"]- sales["costs"]
fig, ax = plt.subplots()
ax.set_xlabel("months")
ax.set_ylabel("revenue")
ax.plot(sales["month"],sales["profit"])
plt.show()
```

```python
import pandas as pd
import numpy as np


data = {
   "name": [" ananya rao", "RAHUL Verma", "Priya Nair", "Priya Nair", "sana khan", "Vikram  Singh", "Karan Mehta", "meera IYER", "Arjun Das", None,],
   "city": ["Bengaluru", "mumbai", "Chennai", "Chennai", "PUNE", "bengaluru", "Mumbai", "chennai", "Pune", "Bengaluru",],
   "math_score": [78, 55, 92, 93, 20, 64, 71, 95,35,49]
}


raw_students = pd.DataFrame(data)
print(raw_students)
```

```python
#bar chart compare average math score across cities
city_avg = raw_students.groupby("city")["math_score"].mean()
print(city_avg)

fig, ax = plt.subplots()
ax.bar(city_avg.index,city_avg.values)
plt.show()
```

```python
fig,ax = plt.subplots()
ax.plot(sales["month"],sales["revenue"],label="revenue",marker="o",color="tab:blue")
ax.plot(sales["month"],sales["costs"],label="costs",marker="x",color="tab:red")

ax.set_xlabel("month")
ax.set_ylabel("amount")
ax.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "March", "April"]

sales = [100, 150, 130, 200]
costs = [70, 90, 100, 120]

fig, ax = plt.subplots()

ax.plot(months, sales, label="Sales")
ax.plot(months, costs, label="Costs")

ax.set_title("Monthly Sales and Costs")
ax.set_xlabel("Month")
ax.set_ylabel("Amount (Rs)")

ax.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "March", "April"]

sales = [100, 150, 130, 200]
costs = [70, 90, 100, 120]

fig,ax = plt.subplots()
ax.plot(months,sales,label="sales",color="green",linewidth=2)
ax.plot(months,costs,label="costs",color="red",marker="o")
ax.set_xlabel("Month")
ax.set_ylabel("amount(rs)")
ax.set_title("Revenue vs costs")
ax.legend()
plt.show()
```
