```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


sales = pd.DataFrame({
   "month": ["Jan", "Feb", "March", "Apr", "May", "June"],
   "revenue": [120, 135, 128, 160, 175, 190],
   "costs": [90, 95, 100, 110, 120, 125]
})


print(sales)
```

```python
fig,ax = plt.subplots()
ax.plot(sales["month"],sales["revenue"])
plt.show()
```

```python
# add a profit column (revenue - costs) to sales and draw it as a  line chart
sales ["profit"] = sales["revenue"]- sales["costs"]
fig, ax = plt.subplots()
ax.plot(sales["month"],sales["profit"])
plt.show()
```

```python
sales ["profit"] = sales["revenue"]- sales["costs"]
fig, ax = plt.subplots()
ax.set_xlabel("months")
ax.set_ylabel("revenue")
ax.plot(sales["month"],sales["profit"])
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "study_hours": [1, 2, 3, 4, 5],
    "marks": [35, 40, 55, 70, 85],
    "sleep_hours": [8, 7, 8, 6, 7]
}

df = pd.DataFrame(data)
correlation = df.corr()
fig,ax=plt.subplots()
sns.heatmap(correlation, annot=True, ax=ax)
ax.set_title("Correlation Heatmap")
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "March", "Apr"]
revenue = [100, 120, 150, 180]

fig, ax = plt.subplots()

ax.plot(months, revenue)
ax.set_title("Monthly Revenue")
ax.set_xlabel("Month")
ax.set_ylabel("Revenue")
fig.savefig("sales.png")
```
